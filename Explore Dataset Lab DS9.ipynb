Print the mean age of the survey participants.


```python
## Write your code here
au = df["Age"].unique()
print({"Number of Unique Age Range":len(au)})
print(au)

auc = df["Age"].value_counts()
print(auc)

dfa = pd.DataFrame(auc)
age_range = dfa.rename(index = {"25-34 years old":29.5,
                                   "35-44 years old":39.5,
                                   "18-24 years old":21,
                                   "45-54 years old":49.5,
                                   "55-64 years old":59.5,
                                   "Under 18 years old":9,
                                   "65 years or older":82.5,
                                   "Prefer not to say":40})
print(age_range)

dfs = pd.DataFrame([[29.5,23913,29.5*23913],
                    [39.5,14948,39.5*14948],
                    [21.0,14105,21.0*14105],
                    [49.5,6251,49.5*6251],
                    [59.5,2575,59.5*2575],
                    [9.0,2570,9.0*2570],
                    [82.5,773,82.5*773],
                    [40.0,322,40.0*322]],
                  columns=['Midpoint','Count','Sum'])
print(dfs)

mean_age = dfs['Sum'].mean()
print("Mean Age of Participants =",mean_age*.0001)
```
